```python
import re

def clean_tunisian_text(text):
    if pd.isna(text):
        return ""
    text = str(text).lower().strip()
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r"[إأآ]", "ا", text)
    text = re.sub(r"ة", "ه", text)
    text = re.sub(r"ى", "ي", text)
    return text

df['Question_Clean'] = df['Question'].apply(clean_tunisian_text)
print("✅ Nettoyage terminé !")
print(df[['Question', 'Question_Clean']].head())
```

```python
def chatbot_vocal_logic(user_input):
    query_clean = clean_tunisian_text(user_input)
    query_embedding = model.encode(query_clean, convert_to_tensor=True)
    cosine_scores = util.cos_sim(query_embedding, question_embeddings)
    top_result = torch.topk(cosine_scores, k=1)
    score = top_result.values[0].item()
    index = top_result.indices[0].item()
    print(f"--- [Score de confiance : {score:.2f}] ---")
    if score > 0.4:
        return df['Response'].iloc[index]
    else:
        return "Manich fehem klemik blegda. Najem n'awnek fi aswem el ticket walla el awaket?"
```

```python
import gradio as gr
from gtts import gTTS
import tempfile

def chatbot_pipeline(audio):
    try:
        # 1. Transcription Whisper
        result = stt_model.transcribe(audio, language="ar")
        user_text = result["text"].strip()
        print(f"🗣️ Vous : {user_text}")

        if not user_text:
            return "⚠️ Aucun son détecté", None

        # 2. Réponse chatbot
        bot_response_text = chatbot_vocal_logic(user_text)
        print(f"🤖 Bot : {bot_response_text}")

        # 3. Synthèse vocale
        tts = gTTS(text=bot_response_text, lang='ar', slow=False)
        tmp = tempfile.NamedTemporaryFile(suffix=".mp3", delete=False)
        tts.save(tmp.name)

        return bot_response_text, tmp.name

    except Exception as e:
        print(f"❌ Erreur : {e}")
        return f"Erreur : {e}", None

# Interface Gradio
interface = gr.Interface(
    fn=chatbot_pipeline,
    inputs=gr.Audio(sources="microphone", type="filepath", label="🎤 Parle ici"),
    outputs=[
        gr.Textbox(label="🤖 Réponse du bot"),
        gr.Audio(label="🔊 Réponse vocale", autoplay=True)
    ],
    title="🚌 Chatbot Transport Tunisien",
    description="Parle en arabe ou en tunisien — le bot te répond à voix haute !",
    live=False
)

interface.launch(share=True)
```
